## 3: FEATURE ENGINEERING

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(
    "../data/cleaned/attendance_cleaned.csv"
)
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df["Day of Semester"] = (
    df["Date"] - df["Date"].min()
).dt.days + 1
```

```python
df["Week Number"] = (
    ((df["Date"] - df["Date"].min()).dt.days // 7) + 1
)
```

```python
df["Start Time"] = pd.to_datetime(
    df["Start Time"],
    format="%H:%M",
    errors="coerce"
)
```

```python
df["Time Period"] = np.where(
    df["Start Time"].dt.hour < 12,
    "Morning",
    "Afternoon"
)
```

```python
df = df.sort_values(
    ["Date", "Lecture Number"]
)
```

```python
df["Rolling 3 Lecture Attendance"] = (
    df["Attendance %"]
    .shift(1)
    .rolling(3)
    .mean()
)
```

```python
binary_columns = [
    "Test Week",
    "Assignment Due",
    "Holiday",
    "Special Event"
]

for col in binary_columns:
    df[col] = df[col].map({
        "Yes": 1,
        "No": 0
    })
```

```python
def attendance_category(x):
    if x < 50:
        return "Low"
    elif x <= 75:
        return "Medium"
    else:
        return "High"

df["Attendance Category"] = (
    df["Attendance %"]
    .apply(attendance_category)
)
```

```python
df.to_csv(
    "../data/cleaned/attendance_engineered.csv",
    index=False
)
```
