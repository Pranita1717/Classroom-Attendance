## 2. Data Cleaning


```python
import pandas as pd
import numpy as np

df = pd.read_excel("D:/Classroom Attendance/data/raw/MCA_Attendance_Dataset_Faculty_ID_Encoded_300_Rows.xlsx")
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df["Date"].dtype
```

```python
df.duplicated().sum()
```

```python
df["Calculated Attendance %"] = (
    df["Students Present"] /
    df["Total Enrolled Students"]
) * 100
```

```python
df[
    ["Students Present",
     "Total Enrolled Students",
     "Attendance %",
     "Calculated Attendance %"]
].head()
```

```python
df[
    (df["Attendance %"] < 0) |
    (df["Attendance %"] > 100)
]
```

```python
df[
    df["Students Present"] >
    df["Total Enrolled Students"]
]
```

```python
df["Faculty Experience Years"] = (
    df["Faculty Experience"]
    .astype(str)
    .str.extract(r"(\d+)")
    .astype(float)
)
```

```python
df[
    ["Faculty Experience",
     "Faculty Experience Years"]
].head()
```

```python
df["Previous Lecture Attendance"].isnull().sum()
```

```python
df["Previous Attendance Missing"] = (
    df["Previous Lecture Attendance"]
    .isna()
    .astype(int)
)
```

```python
df["Previous Lecture Attendance"] = (
    df["Previous Lecture Attendance"]
    .fillna(df["Attendance %"].mean())
)
```

```python
df.drop(
    columns=["Calculated Attendance %"],
    inplace=True
)
```

```python
df.to_csv(
    "../data/cleaned/attendance_cleaned.csv",
    index=False
)
```
