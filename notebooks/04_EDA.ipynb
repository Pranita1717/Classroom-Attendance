## 4. EDA

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(
    "../data/cleaned/attendance_engineered.csv"
)
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df["Attendance %"],
    bins=10,
    kde=True
)

plt.title("Distribution of Attendance Percentage")
plt.xlabel("Attendance %")
plt.ylabel("Number of Lectures")

plt.show()
```

```python
subject_attendance = (
    df.groupby("Subject")["Attendance %"]
    .mean()
    .sort_values()
)

subject_attendance.plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("Average Attendance by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Attendance %")
plt.xticks(rotation=45)

plt.show()
```

```python
day_attendance = (
    df.groupby("Day")["Attendance %"]
    .mean()
)

day_attendance.plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Average Attendance by Day")
plt.ylabel("Average Attendance %")

plt.show()
```

```python
lecture_attendance = (
    df.groupby("Lecture Number")["Attendance %"]
    .mean()
)

lecture_attendance.plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Average Attendance by Lecture Number")
plt.xlabel("Lecture Number")
plt.ylabel("Average Attendance %")

plt.show()
```

```python
sns.boxplot(
    data=df,
    x="Practical/Theory",
    y="Attendance %"
)

plt.title("Practical vs Theory Attendance")
plt.show()
```

```python
sns.boxplot(
    data=df,
    x="Test Week",
    y="Attendance %"
)

plt.title("Attendance During Test Week")
plt.show()
```

```python
sns.boxplot(
    data=df,
    x="Weather",
    y="Attendance %"
)

plt.title("Attendance by Weather")
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="Previous Lecture Attendance",
    y="Attendance %"
)

plt.title(
    "Previous Lecture Attendance vs Current Attendance"
)

plt.show()
```
